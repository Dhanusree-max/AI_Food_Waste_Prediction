```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/food_waste_master_dataset.csv")

df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Inventory_Turnover"] = (
    df["Inventory_Sold"] / df["Inventory_Received"]
)
```

```python
df[["Inventory_Received",
    "Inventory_Sold",
    "Inventory_Turnover"]].head()
```

## Observation

Inventory Turnover measures the proportion of inventory sold.

Higher values indicate efficient inventory movement, while lower values may indicate excess stock and a higher risk of food waste.

```python
# Calculate the percentage of inventory wasted

df["Waste_Percentage"] = (
    df["Waste_Quantity"] / df["Inventory_Received"]
) * 100
```

```python
df[[
    "Inventory_Received",
    "Waste_Quantity",
    "Waste_Percentage"
]].head()
```

## Observation

Waste Percentage represents the proportion of received inventory that was wasted.

This feature provides a standardized measure of waste, allowing fair comparison across products with different inventory levels.

### Business Insight

Products with a high waste percentage should be prioritized for inventory optimization, pricing strategies, or promotional campaigns to reduce losses.

Sales Velocity = Inventory Sold / Shelf Life Days

```python
# Calculate the average number of units sold per shelf-life day

df["Sales_Velocity"] = (
    df["Inventory_Sold"] / df["Shelf_Life_Days"]
)
```

```python
df[[
    "Inventory_Sold",
    "Shelf_Life_Days",
    "Sales_Velocity"
]].head()
```

## Observation

Sales Velocity measures how quickly products are sold relative to their shelf life.

Higher values indicate faster product movement, reducing the likelihood of expiry and food waste.

### Business Insight

Products with low sales velocity should be monitored closely. Businesses can improve sales through discounts, promotions, or stock transfers before products expire.

Stock Utilization=Inventory Sold/Inventory Sold+Inventory Remaining

```python
# Calculate stock utilization

df["Stock_Utilization"] = (
    df["Inventory_Sold"] /
    (df["Inventory_Sold"] + df["Inventory_Remaining"])
)
```

```python
df[[
    "Inventory_Sold",
    "Inventory_Remaining",
    "Stock_Utilization"
]].head()
```

## Observation

Stock Utilization measures how efficiently the available inventory was utilized.

Higher values indicate better inventory management, while lower values suggest excess inventory and potential waste.

### Business Insight

Stores with consistently low stock utilization may require improved demand forecasting or reduced inventory levels to minimize food waste.

```python
# Create expiry risk category

df["Expiry_Risk"] = np.where(
    df["Expiry_Days_Left"] <= 7,
    "High",
    np.where(
        df["Expiry_Days_Left"] <= 30,
        "Medium",
        "Low"
    )
)
```

```python
df[[
    "Expiry_Days_Left",
    "Expiry_Risk"
]].head(10)
```

## Observation

Products are categorized into High, Medium, and Low expiry risk based on the remaining shelf life.

This feature helps identify products that require immediate attention.

### Business Insight

Products with High Expiry Risk should be prioritized for discounts, promotions, or stock transfers to reduce waste.

```python
# Create high temperature flag

df["High_Temperature_Flag"] = np.where(
    df["Temperature"] > 30,
    1,
    0
)
```

```python
df[[
    "Temperature",
    "High_Temperature_Flag"
]].head(10)
```

## Observation

The High Temperature Flag identifies records where the temperature exceeds 30°C.

This feature can help the model learn the impact of high temperatures on food spoilage.

### Business Insight

During periods of high temperature, businesses can strengthen cold-chain monitoring, increase refrigeration checks, and adjust inventory levels for temperature-sensitive products.
