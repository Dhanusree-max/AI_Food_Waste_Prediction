```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/food_waste_feature_engineered.csv")

df.head()
```

```python
df.columns
```

```python
y = df["Waste_Quantity"]
```

```python
columns_to_drop = [
    "Store_ID",
    "Product_ID",
    "Product_Name",
    "Store_Name",
    "Date",
    "Waste_Quantity",          # Target
    "Waste_Cost",              # Calculated from Waste_Quantity
    "Waste_Risk",              # Derived from Waste_Quantity
    "Inventory_Recommendation" # Depends on Waste_Quantity
]

#Now we need to remove columns that should not be used for training.
#Why?
#Some columns either:
#Are IDs (they don't help prediction).
#Directly reveal the answer (data leakage).
```

```python
X = df.drop(columns=columns_to_drop)
```

```python
print(X.shape)
print(y.shape)
```

## Observation

The target variable (`Waste_Quantity`) has been separated from the input features.

Identifier columns and variables that could cause data leakage were removed to ensure the model learns meaningful patterns rather than relying on information derived from the target.

```python
X.select_dtypes(include="object").columns
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
```

```python
categorical_columns = X.select_dtypes(include="object").columns

for column in categorical_columns:
    X[column] = label_encoder.fit_transform(X[column])
```

```python
X.head()
```

## Observation

All categorical features have been converted into numerical values using Label Encoding.

This transformation makes the dataset compatible with machine learning algorithms while preserving the distinct categories.

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

categorical_columns = X.select_dtypes(include="object").columns

for column in categorical_columns:
    X[column] = label_encoder.fit_transform(X[column])

X.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)
print("Training Target   :", y_train.shape)
print("Testing Target    :", y_test.shape)
```

## Observation

- The dataset has been divided into training and testing sets.
- 80% of the data is used to train the machine learning model.
- 20% of the data is reserved for evaluating the model on unseen data.
- Using a fixed `random_state` ensures that the split is reproducible.

```python
X.to_csv("../data/processed/X_preprocessed.csv", index=False)
y.to_csv("../data/processed/y_target.csv", index=False)

print("Preprocessed data saved successfully!")
```
